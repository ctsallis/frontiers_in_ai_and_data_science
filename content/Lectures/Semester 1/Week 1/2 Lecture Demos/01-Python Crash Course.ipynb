```python
1 / 2
```

```python
[1,2,3]
```

```python
['hi',1,[1,2]]
```

```python
my_list = ['a','b','c']
```

```python
my_list.append('d')
```

```python
my_list
```

```python
my_list[0]
```

```python
my_list[1]
```

```python
my_list[1:]
```

```python
my_list[:1]
```

```python
my_list[0] = 'NEW'
```

```python
my_list
```

Lists can contain lists. Chain the indexes from the outside in: `nest[3]` is the inner list, `nest[3][2]` is the list inside that, and `nest[3][2][0]` is the string inside it.

```python
nest = [1,2,3,[4,5,['target']]]
```

```python
nest[3]
```

```python
nest[3][2]
```

```python
nest[3][2][0]
```

## if, elif, else Statements

`if` runs the indented block only when its condition is `True`. `else` covers every other case, and `elif` ("else if") checks another condition in between. The indentation (4 spaces) tells Python which lines belong to which branch.

```python
if 1 < 2:
    print('Yep!')
```

```python
if 1 < 2:
    print('yep!')
```

```python
if 1 < 2:
    print('first')
else:
    print('last')
```

```python
if 1 > 2:
    print('first')
else:
    print('last')
```

```python
if 1 == 2:
    print('first')
elif 3 == 3:
    print('middle')
else:
    print('Last')
```

## for Loops

A `for` loop repeats its indented block once for each item in a sequence. The loop variable (`item`, `jelly`, ...) can have any name; it holds the current item.

```python
seq = [1,2,3,4,5]
```

```python
for item in seq:
    print(item)
```

```python
for item in seq:
    print('Yep')
```

```python
for jelly in seq:
    print(jelly+jelly)
```

## while Loops

A `while` loop repeats as long as its condition stays `True`. Here `i` grows by 1 each time, so the loop stops when `i` reaches 5; without `i = i+1` it would never end.

```python
i = 1
while i < 5:
    print('i is: {}'.format(i))
    i = i+1
```

## range()

`range(5)` produces the numbers 0, 1, 2, 3, 4: it starts at 0 and stops before 5. It produces them one at a time, so wrap it in `list()` to see them all at once.

```python
range(5)
```

```python
for i in range(5):
    print(i)
```

```python
list(range(5))
```

## list comprehension

A list comprehension builds a new list in one line. The two cells below give the same result: first with a `for` loop and `.append()`, then as `[expression for item in sequence]`.

```python
x = [1,2,3,4]
```

```python
out = []
for item in x:
    out.append(item**2)
print(out)
```

```python
[item**2 for item in x]
```

</details>

## functions

`def` creates a reusable function. `param1='default'` is a default value, used when you call the function without an argument. The text in triple quotes (the *docstring*) describes what the function does. Typing the name without brackets only shows the function; adding `()` runs it.

```python
def my_func(param1='default'):
    """
    Docstring goes here.
    """
    print(param1)
```

```python
my_func
```

```python
my_func()
```

```python
my_func('new param')
```

```python
my_func(param1='new param')
```

`return` sends a value back, so you can store it, as in `out = square(2)`. A function that only prints (like `my_func`) gives nothing back.

```python
def square(x):
    return x**2
```

```python
out = square(2)
```

```python
print(out)
```

## lambda expressions

A lambda is a short, unnamed function written on one line: `lambda var: var*2` does the same job as `times2`. It is useful when a function is needed only once, e.g. inside `map` or `filter`.

```python
def times2(var):
    return var*2
```

```python
times2(2)
```

```python
lambda var: var*2
```

## map and filter

`map(f, seq)` applies the function `f` to every item; `filter(f, seq)` keeps only the items for which `f` returns `True`.

```python
seq = [1,2,3,4,5]
```

```python
map(times2,seq)
```

```python
list(map(times2,seq))
```

```python
list(map(lambda var: var*2,seq))
```

```python
filter(lambda item: item%2 == 0,seq)
```

```python
list(filter(lambda item: item%2 == 0,seq))
```

**What do you notice?** On its own, `map(...)` or `filter(...)` shows something like `<map at 0x...>` instead of a list.

<details><summary>Answer</summary>

`map` and `filter` are lazy: they produce their values only when asked. Wrapping them in `list()` collects all the values. The `0x...` part is a memory address and changes every time.

</details>
