**Create a function that grabs the email website domain from a string in the form:**

    user@domain.com
    
**So for example, passing "user@domain.com" would return: domain.com**

```python
def domainGet(email):
    # split at '@' and keep the last part
    return email.split('@')[-1]
```

```python
domainGet('user@domain.com')
```

**Create a basic function that returns True if the word 'dog' is contained in the input string. Don't worry about edge cases like punctuation attached to the word dog, but do account for capitalization.**

```python
def findDog(st):
    # lower() ignores capitals; split() turns the sentence into a list of words
    return 'dog' in st.lower().split()
```

```python
findDog('Is there a dog here?')
```

**Create a function that counts the number of times the word "dog" occurs in a string. Again ignore edge cases.**

```python
def countDog(st):
    # go through the words one by one and count those equal to 'dog'
    count = 0
    for word in st.lower().split():
        if word == 'dog':
            count += 1
    return count
```

```python
countDog('This dog runs faster than the other dog dude!')
```

**You are driving a little too fast, and a police officer stops you. Write a function
  to return one of 3 possible results: "No Ticket", "Small Ticket", or "Big Ticket". 
  If your speed is 60 or less, the result is "No Ticket". If speed is between 61 
  and 80 inclusive, the result is "Small Ticket". If speed is 81 or more, the result is "Big Ticket". Unless it is your birthday (encoded as a boolean value in the parameters of the function) -- on your birthday, your speed can be 5 higher in all 
  cases.**

```python
def caught_speeding(speed, is_birthday):
    
    # on your birthday every limit is 5 higher, which is the same as testing speed - 5
    if is_birthday:
        speeding = speed - 5
    else:
        speeding = speed
    
    if speeding > 80:
        return 'Big Ticket'
    elif speeding > 60:
        return 'Small Ticket'
    else:
        return 'No Ticket'
```

```python
caught_speeding(81,True)
```

```python
caught_speeding(81,False)
```
