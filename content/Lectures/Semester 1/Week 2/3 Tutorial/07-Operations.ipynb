## Useful operations

There are lots of operations with pandas that will be really useful to you, but don't fall into any distinct category. Let's show them here in this lecture:

```python
import pandas as pd
df = pd.DataFrame({'col1':[1,2,3,4],'col2':[444,555,666,444],'col3':['abc','def','ghi','xyz']})
df.head()
```

### Info on Unique Values

`unique` lists the distinct values, `nunique` counts them, and `value_counts` counts how often each value occurs, most frequent first. On the course website `unique()` may also show `dtype=int64` after the list, because the browser version of Python uses 32-bit whole numbers by default; the values are the same.

```python
df['col2'].unique()
```

```python
df['col2'].nunique()
```

```python
df['col2'].value_counts()
```

### Selecting Data

```python
#Select from DataFrame using criteria from multiple columns
newdf = df[(df['col1']>2) & (df['col2']==444)]
```

```python
newdf
```

### Applying Functions

`apply` calls a function on every value of a column and returns the results as a new Series. Any function that takes one value works, your own (`times2`) or a built-in one such as `len`.

```python
def times2(x):
    return x*2
```

```python
# apply times2 to every value of col1
df['col1'].apply(times2)
```

```python
df['col3'].apply(len)
```

```python
df['col1'].sum()
```

`del` changes `df` itself, unlike `drop` without `inplace=True`.

**Find null values or check for null values**

`isnull()` marks each missing value with True. This table has none, so `dropna()` returns it unchanged.

```python
df.isnull()
```

```python
# Drop rows with NaN Values
df.dropna()
```

**Filling in NaN values with something else:**

Filling with text turns the number columns into general Python objects, so their numbers are shown as `1.0`, `555.0` and so on.

```python
import numpy as np
```

```python
df = pd.DataFrame({'col1':[1,2,3,np.nan],
                   'col2':[np.nan,555,666,444],
                   'col3':['abc','def','ghi','xyz']})
df.head()
```

```python
df.fillna('FILL')
```

**Pivot tables:** `pivot_table` summarises column `D` with one row for each combination of `A` and `B` and one column for each value of `C`. Each cell is the mean of the matching `D` values (here each combination occurs at most once); `NaN` means the combination does not occur.

```python
data = {'A':['foo','foo','foo','bar','bar','bar'],
     'B':['one','one','two','two','one','one'],
       'C':['x','y','x','y','x','y'],
       'D':[1,3,2,5,4,1]}

df = pd.DataFrame(data)
```

```python
df
```

```python
df.pivot_table(values='D',index=['A', 'B'],columns=['C'])
```

**Try it:** add `aggfunc='sum', fill_value=0` to the `pivot_table` call. What changes?
