## Methods

Let's show a few convenient methods to deal with Missing Data in pandas. The example table has missing values (`np.nan`) in columns A and B; column C is complete.

```python
import numpy as np
import pandas as pd
```

```python
df = pd.DataFrame({'A':[1,2,np.nan],
                  'B':[5,np.nan,np.nan],
                  'C':[1,2,3]})
```

```python
df
```

`dropna` removes data and `fillna` replaces it. Neither changes `df` itself: each returns a new table.

```python
# drop every row that has at least one missing value
df.dropna()
```

```python
# drop every column that has a missing value
df.dropna(axis=1)
```

```python
# keep only the rows with at least 2 values that are not missing
df.dropna(thresh=2)
```

Filling with text turns columns A and B into general Python objects, so their numbers are shown as `1.0` and `5.0` next to the text. In practice you fill a number column with a number, as in the next cell.

```python
# replace every missing value with the same constant
df.fillna(value='FILL VALUE')
```

```python
# fill the gap in column A with the mean of its known values: (1 + 2) / 2 = 1.5
df['A'].fillna(value=df['A'].mean())
```
