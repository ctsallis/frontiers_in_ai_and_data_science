# Introduction to pandas (Week 2 tutorial)

**What this set of notebooks covers:** pandas, the Python library for working with tables of data: creating and selecting data, handling missing values, grouping, combining tables, and reading and writing files. These are the tools behind the data cleaning, integration and transformation steps in this week's lecture, *1.2 Data Preprocessing*.

**How to run them:** the notebooks run in your browser, so there is nothing to install. Open a notebook, click a code cell and press **Shift + Enter** to run it and move to the next one, or use **Run → Run All Cells**. Run the cells in order, because later cells use variables created earlier.

**What to expect:** each code cell shows its result underneath, usually a small table. The saved results are already there, so you can read a notebook before running it.
