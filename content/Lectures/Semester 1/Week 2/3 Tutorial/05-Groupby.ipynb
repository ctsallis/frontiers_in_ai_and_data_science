## Grouping rows

The groupby method allows you to group rows of data together and call aggregate functions.

```python
import pandas as pd
# Create dataframe: one row per salesperson
data = {'Company':['GOOG','GOOG','MSFT','MSFT','FB','FB'],
       'Person':['Sam','Charlie','Amy','Vanessa','Carl','Sarah'],
       'Sales':[200,120,340,124,243,350]}
```

```python
df = pd.DataFrame(data)
```

```python
df
```

**Now you can use the .groupby() method to group rows together based off of a column name. For instance let's group based off of Company. This will create a DataFrameGroupBy object:**

On its own, the object only shows its type and memory address. It holds the groups until you ask for a summary.

```python
df.groupby('Company')
```

You can save this object as a new variable:

```python
by_comp = df.groupby("Company")
```

And then call aggregate methods off the object. `numeric_only=True` tells pandas to summarise only the number columns: the mean of the names in `Person` cannot be computed, and current pandas versions stop with a `TypeError` without this option.

```python
by_comp.mean(numeric_only=True)
```

```python
df.groupby('Company').mean(numeric_only=True)
```

More examples of aggregate methods. `min` and `max` also work on text (in alphabetical order), and `count` counts the values that are not missing in each column.

```python
by_comp.std(numeric_only=True)
```

```python
by_comp.min()
```

```python
by_comp.max()
```

```python
by_comp.count()
```

`describe()` gives the count, mean, standard deviation, minimum, quartiles and maximum of each group at once. `transpose()` swaps rows and columns, and `['GOOG']` then picks one company.

```python
by_comp.describe()
```

```python
by_comp.describe().transpose()
```

```python
by_comp.describe().transpose()['GOOG']
```
