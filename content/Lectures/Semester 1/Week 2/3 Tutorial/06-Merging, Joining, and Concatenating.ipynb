## Concatenation

Concatenation basically glues together DataFrames. Keep in mind that dimensions should match along the axis you are concatenating on. You can use **pd.concat** and pass in a list of DataFrames to concatenate together. With `axis=1` the tables are placed side by side instead and rows are matched by index label; no label appears in two of these tables, so most cells are `NaN`.

```python
pd.concat([df1,df2,df3])
```

```python
pd.concat([df1,df2,df3],axis=1)
```

## Merging

The **merge** function allows you to merge DataFrames together using a similar logic as merging SQL Tables together. `how='inner'` keeps only the keys found in both tables. For example:

```python
pd.merge(left,right,how='inner',on='key')
```

Or to show a more complicated example. Here rows must match on both `key1` and `key2`, and `how` decides which unmatched rows are kept: `'outer'` keeps all of them, `'left'` keeps every row of `left`, and `'right'` every row of `right`. The key pair (K1, K0) appears twice in `right`, so it produces two rows.

```python
left = pd.DataFrame({'key1': ['K0', 'K0', 'K1', 'K2'],
                     'key2': ['K0', 'K1', 'K0', 'K1'],
                        'A': ['A0', 'A1', 'A2', 'A3'],
                        'B': ['B0', 'B1', 'B2', 'B3']})
    
right = pd.DataFrame({'key1': ['K0', 'K1', 'K1', 'K2'],
                               'key2': ['K0', 'K0', 'K0', 'K0'],
                                  'C': ['C0', 'C1', 'C2', 'C3'],
                                  'D': ['D0', 'D1', 'D2', 'D3']})
```

```python
# inner merge (the default): only key pairs found in both tables
pd.merge(left, right, on=['key1', 'key2'])
```

```python
# keep every key pair from both tables; values with no match become NaN
pd.merge(left, right, how='outer', on=['key1', 'key2'])
```

```python
# keep every key pair of the right table
pd.merge(left, right, how='right', on=['key1', 'key2'])
```

```python
# keep every key pair of the left table
pd.merge(left, right, how='left', on=['key1', 'key2'])
```

## Joining
Joining is a convenient method for combining the columns of two potentially differently-indexed DataFrames into a single result DataFrame. By default `join` keeps every row of the left table (like `how='left'`).

```python
left = pd.DataFrame({'A': ['A0', 'A1', 'A2'],
                     'B': ['B0', 'B1', 'B2']},
                      index=['K0', 'K1', 'K2']) 

right = pd.DataFrame({'C': ['C0', 'C2', 'C3'],
                    'D': ['D0', 'D2', 'D3']},
                      index=['K0', 'K2', 'K3'])
```

```python
left.join(right)
```

```python
left.join(right, how='outer')
```

**Try it:** what does `left.join(right, how='inner')` return?
